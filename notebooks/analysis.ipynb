##### Instead of using the median of overall dataset, using the median of the specific category of items that were purchased


```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

#### Changing column names into snake (small)

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ','_')
```

#### Creating new column for age groups

```python
labels = ['Young','Adult','Middle_Aged','Senior'] 
```

```python
df['age_group'] = pd.qcut(df['age'],q=4,labels=labels) #qcut is used to divide the data into 4 quartiles according to the age col and labels are assigned accordingly
```

```python
df[['age_group','age']].head()
```

#### Conerting the textual frequncy days into numbers

```python
df['frequency_of_purchases'].unique()
```

```python
frequncy ={
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Annually' : 365,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Monthly' : 30,
    'Every 3 Months' : 90
}
```

```python
df['frequency_of_purchases'] = df['frequency_of_purchases'].map(frequncy)
```

#### connecting the notebook with the postgresql
